```python
# Rename Column Names such as "Phone_Number to Contact_Number" / "Do_Not_Contact to Block_Contact"
df = df.rename(columns={'Phone_Number':'Contact_Number', 'Do_Not_Contact':'Block_Contact'})
df
```

```python
# make lowercase of first_name column in dataset 
df['First_Name'].str.lower()
```

```python
# make uppercase of 'Address' column in dataset
df['Address'].str.upper()
```

```python
# setting prticular column as index
df.set_index('CustomerID')
```

```python
# Remove CustomerID column as index
df = df.reset_index(drop=True)
df
```

```python
# Now Lets clean the rows data starting from First_Name and Last_Name. Clean all the special characters and spaces from both columns.
df['First_Name'] = df['First_Name'].str.replace('/','')
df['Last_Name'] = df['Last_Name'].str.replace('/','')
df
```

```python
# Fix: The error "ValueError: invalid literal for int() with base 10: '123-545-5421'" occurred because the column contains non-numeric characters.
# We will remove the non-numeric characters and then convert the column to a numeric type that can handle missing values.
df['Contact_Number'] = df['Contact_Number'].str.replace('[^0-9]', '', regex=True)
df['Contact_Number'] = pd.to_numeric(df['Contact_Number'], errors='coerce').astype('Int64')
df.dtypes
```

```python
# Remove all the special characters "/,|"" from Contact_Number.
df["Contact_Number"] = df["Contact_Number"].astype(str).str.replace(r'[^0-9-]', '', regex=True)
df
```

```python
# In column contact number add "-" after every 3 numbers.
df['Contact_Number'] = df['Contact_Number'].str[:3] +'-'+ df['Contact_Number'].str[3:6] + '-' + df['Contact_Number'].str[6:]
df
```

```python
# In column Contact_Number remove "--".
df['Contact_Number']=df['Contact_Number'].str.replace('--','')
df
```

```python
# In Column Address remove all "," and replace with ''.
df['Address'] = df['Address'].str.replace(',', '')
df
```

```python
# In Column Address remove all "N/a" and replace with ''.
df['Address'] = df['Address'].str.replace('N/a','')
df
```

```python
# From Columns "Paying Customer" find data with "Yes" and replace with "Y" and "No" with "N"
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes','Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No','N')
df['Paying Customer'] = df['Paying Customer'].str.replace('N/a','')
df
```

```python
# From Columns "Block_Contact" find data with "Yes" and replace with "Y" and "No" with "N"
df['Block_Contact'] = df['Block_Contact'].str.replace('Yes','Y')
df['Block_Contact'] = df['Block_Contact'].str.replace('No','N')
df
```

```python
# From Columns "Not_Useful_Column" find data with "True" and replace with "T" and "False" with "F"
# Fix: The error "AttributeError: Can only use .str accessor with string values!" occurred because the column is not of string data type.
# We will convert the column to string data type before applying the replace method.
```

```python
df['Not_Useful_Column'].dtype
```

```python
df['Not_Useful_Column'] = df['Not_Useful_Column'].astype(str).str.replace('True' , 'T')
df['Not_Useful_Column'] = df['Not_Useful_Column'].astype(str).str.replace('False' , 'F')
df
```

```python
# In column "Block_Contact", display only results with "N"
for x in df.index:
    if df.loc[x, 'Block_Contact'] == 'Y':
        df.drop(x, inplace=True)
df    
```

```python
df.to_excel("customers_cleaned.xlsx", index=False)
```
