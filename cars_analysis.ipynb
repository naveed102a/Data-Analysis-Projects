```python
# hamry pas MSRP ki values string me hy to usko pehly shi krna pryga to mny ek (MSRP_num) ek new column banaya phr $ or , remove 
# krky usko ek integer bana kr usko us column me store kiya 
df['MSRP_num'] = pd.to_numeric(
    df['MSRP'].str.replace('$', '', regex=False)
              .str.replace(',', '', regex=False)
)
df[df['MSRP_num'] > 50000][['Make','Model','Type','MSRP']]
```

### Level 3 (Aggregation & GroupBy)

- Har company ki average horsepower nikalo.

```python
df.groupby('Make')['Horsepower'].mean().sort_values(ascending=False)
```

_____________
- find the average MSRP or every origin

```python
ori = df.groupby('Origin', as_index=False)['MSRP_num'].mean()
ori
```

```python
plt.figure(figsize=(5,3))
sns.barplot(x = ori.index, y = ori.values, palette='viridis', width=0.5)
```

______________
-  Find average MPG_City of Every Type.

```python
avg_mpg = df.groupby('Type')['MPG_City'].mean()
avg_mpg
```

```python
plt.figure(figsize=(5,3))
sns.barplot(x = avg_mpg.index, y = avg_mpg.values, palette='PuOr')
plt.title('Average MPG_City of Every Car type')
```
