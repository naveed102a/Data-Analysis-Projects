```python
plt.plot(emp_data['Salary_usd'], emp_data['Test_score'])
plt.plot(emp_data['Salary_usd'], emp_data['Interview_score'])

plt.title('Employee Salary vs  Test & Interview Score')
plt.xlabel('salary USD')
plt.ylabel('Score')
plt.show()
```

```python
plt.plot(emp_data['Salary_usd'], emp_data['Test_score'], color = '#D9F10F', linestyle = 'solid', linewidth = 3)
plt.plot(emp_data['Salary_usd'], emp_data['Interview_score'], color = '#FC00D6', linestyle = 'dashdot', linewidth = 2)

plt.title('Employee Salary vs  Test & Interview Score')
plt.xlabel('salary USD')
plt.ylabel('Score')
plt.show()
```

```python
plt.plot(emp_data['Salary_usd'], emp_data['Test_score'], color = '#D9F10F', linestyle = 'solid', linewidth = 3, marker = 'D', markersize=10)
plt.plot(emp_data['Salary_usd'], emp_data['Interview_score'], color = '#FC00D6', linestyle = 'dashdot', linewidth = 2, marker = '>')

plt.title('Employee Salary vs  Test & Interview Score')
plt.xlabel('salary USD')
plt.ylabel('Score')
plt.show()
```

```python
plt.plot(emp_data['Salary_usd'], emp_data['Test_score'], color = '#D9F10F', linestyle = 'solid', linewidth = 3, marker = 'D', markersize=10, label = 'Test Score')
plt.plot(emp_data['Salary_usd'], emp_data['Interview_score'], color = '#FC00D6', linestyle = 'dashdot', linewidth = 2, marker = '>', label = 'Intervew Score')

plt.title('Employee Salary vs  Test & Interview Score')
plt.xlabel('salary USD')
plt.ylabel('Score')
plt.legend()
plt.show()
```
