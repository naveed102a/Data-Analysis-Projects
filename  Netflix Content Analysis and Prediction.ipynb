## Feature  Engineering

- (Duration Numeric + Type )
Create Duration_Value from values such as '93 min' and '4 Seasons'. Then create Duration_Type with
values such as Minutes and Seasons

```python
# Extract numeric value
df['Duration_Value'] = df['Duration'].str.extract(r'(\d+)').astype(float)
# Extract duration type
df['Duration_Type'] = df['Duration'].str.extract(r'(min|Season|Seasons)')
# Convert values into proper names
df['Duration_Type'] = df['Duration_Type'].replace({
    'min': 'Minutes',
    'Season': 'Seasons',
    'Seasons': 'Seasons'
})
```

```python
df[['Duration', 'Duration_Value', 'Duration_Type']].head(10)
```

-  Create Content_Age using the current year minus Release_Year. For this project use 2026 as the current
year.

```python
df['Content_Age'] = 2026 - df['Release_year']
```

## ML mini model

```python
# Task 15: Prepare X and y

X = df[['Release_year', 'Rating', 'Duration_Value',
        'Duration_Type', 'Content_Age']]
y = df['Category']
# Handle missing values
X = X.dropna()
y = y.loc[X.index]
# Task 16: Encoding
X = pd.get_dummies(
    X,
    columns=['Rating', 'Duration_Type'],
    drop_first=True
)
# Task 17: Train/Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y)
# Task 18: Logistic Regression
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
# Task 19: Evaluation
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay)
# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
# Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred)
plt.title('Confusion Matrix - Movie vs TV Show')
plt.show()
# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
