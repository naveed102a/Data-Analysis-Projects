### 1. Which team won the most matches?

```python
df.head(2)
```

```python
plt.figure(figsize=(8,3))
match_win = df['match_winner'].value_counts()
sns.barplot(y = match_win.index, x = match_win.values, palette='viridis')
plt.title('Most match win by team')
```

### 10 Best Bowling Figures
- Aur agar aap "Best Bowling Figures" nikalna chaho to aam tor par sab se zyada wickets aur phir sab se kam runs wale records dekhe jate hain.

```python
df.head(3)
```

```python
# first we create the column where we store the wikets extract from 'best_bowling_figure'
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0] ) # create new column and put the wicket value from best figure column
df['highest_wickets'] = df['highest_wickets'].astype(int) # we convert datatype of new column as int 
top2 = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top2
```

```python
sns.barplot(y = top2.index, x = top2.values, palette='Oranges')
```
